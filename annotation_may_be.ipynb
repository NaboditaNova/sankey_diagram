```python
excel_file_path = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\Fraunhofer\combined_summary_full.xlsx' 

# Load the Excel file
df = pd.read_excel(excel_file_path)
df = df.fillna(method='ffill')
# Extract values from the first row (row index 0)
node_label_1 = df.iloc[0, 0]  # Value at position (0,0)
node_label_2 = df.iloc[0, 1]  # Value at position (0,1)
link_value   = df.iloc[0, 2]  # Value at position (0,2)

# Create the nodes part:
nodes = [
    {'label': node_label_1},
    {'label': node_label_2}
]

# Create the link part:
links = [
    {
        'source': node_label_1,
        'target': node_label_2,
        'value': link_value
    }
]

df = df.drop(columns=['Process Name', 'Input Material Amount'])

node_labels = {node_label_1, node_label_2}

for idx, row in df.iterrows():
    src = row[0]  # first column as source
    tgt = row[1]  # second column as target
    val = row[2]  # third column as link value

    # Check and add the source node if not already present
    if src not in node_labels:
        nodes.append({'label': src})
        node_labels.add(src)

    # Check and add the target node if not already present
    if tgt not in node_labels:
        nodes.append({'label': tgt})
        node_labels.add(tgt)

    # Append the link from source to target with the given value
    links.append({
        'source': src,
        'target': tgt,
        'value': val
    })

# Add "Mixed Plastics" node if not present
if "Mixed Plastics" not in node_labels:
    nodes.append({'label': "Mixed Plastics"})
    node_labels.add("Mixed Plastics")

# Prepare Data for Plotly Sankey
node_labels = [node['label'] for node in nodes]  # List of node labels
label_to_index = {label: idx for idx, label in enumerate(node_labels)}  # Map labels to indices

# Convert link information to use node indices rather than labels
source_indices = [label_to_index[link['source']] for link in links]
target_indices = [label_to_index[link['target']] for link in links]
values = [link['value'] for link in links]

# Create the Sankey Diagram with Plotly
fig = go.Figure(data=[go.Sankey(
    arrangement="freeform",
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=node_labels,
    ),
    link=dict(
        source=source_indices,
        target=target_indices,
        value=values,
    )
)])

# Add annotations for the material composition of "Mixed Plastics"
annotations = [
    dict(
        x=0.25,  # Adjust x-position near "Mixed Plastics"
        y=0.5,   # Adjust y-position
        text="<b>Mixed Plastics Composition:</b><br>"
             "Film Material: 45.2%<br>"
             "HDPE: 8.0%<br>"
             "PP: 7.8%<br>"
             "PS: 30.8%<br>"
             "PET: 11.9%",
        showarrow=False,
        font=dict(size=12, color="black"),
        align="left",
        xanchor="right"
    )
]

# Add annotations to the layout
fig.update_layout(
    title_text="Plastic Waste Management - Sankey Diagram",
    font_size=12,
    annotations=annotations
)

# Display the Sankey diagram
fig.show()
```
